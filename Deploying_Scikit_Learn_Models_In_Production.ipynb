# BAB 13 — DEPLOYING SCIKIT-LEARN MODELS IN PRODUCTION

## 13.1 Pengenalan

Model machine learning tidak hanya digunakan untuk eksperimen.

Setelah model selesai dibuat, model dapat digunakan dalam aplikasi.

Contoh alurnya:

Dataset
→ Preprocessing
→ Training
→ Evaluation
→ Save Model
→ Deployment
→ Prediction

Pada bab ini kita akan belajar cara menyimpan model dan menggunakannya kembali.

## 13.2 Menyimpan Model dengan Joblib

Joblib dapat digunakan untuk menyimpan model scikit-learn ke dalam file.

Setelah model disimpan, kita tidak perlu melakukan training ulang setiap kali aplikasi dijalankan.

```python
import joblib

from sklearn.datasets import load_iris
from sklearn.ensemble import RandomForestClassifier

data = load_iris()

X = data.data
y = data.target

model = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

model.fit(X, y)

joblib.dump(
    model,
    "iris_model.pkl"
)

print("Model berhasil disimpan.")
```

## 13.3 Load Model

Model yang sudah disimpan dapat dipanggil kembali menggunakan joblib.load().

```python
import joblib

model = joblib.load(
    "iris_model.pkl"
)

print("Model berhasil dimuat.")
```

## 13.4 Melakukan Prediksi

Setelah model dimuat, kita dapat langsung menggunakannya untuk melakukan prediksi.

Data input harus memiliki format fitur yang sama seperti ketika model dilatih.

```python
import joblib

model = joblib.load(
    "iris_model.pkl"
)

sample = [[5.1, 3.5, 1.4, 0.2]]

prediction = model.predict(sample)

print("Prediksi:", prediction)
```

## 13.5 Menyimpan Pipeline

Dalam production, sebaiknya preprocessing dan model disimpan sebagai satu Pipeline.

Contohnya:

Input
→ StandardScaler
→ Logistic Regression
→ Prediction

Dengan Pipeline, preprocessing yang digunakan saat training akan tetap digunakan saat prediction.

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.datasets import load_iris
import joblib

data = load_iris()

X = data.data
y = data.target

pipeline = Pipeline([
    ("scaler", StandardScaler()),
    ("model", LogisticRegression(max_iter=1000))
])

pipeline.fit(X, y)

joblib.dump(
    pipeline,
    "iris_pipeline.pkl"
)

print("Pipeline berhasil disimpan.")
```

## 13.6 Load Pipeline dan Prediction

```python
import joblib

pipeline = joblib.load(
    "iris_pipeline.pkl"
)

sample = [[5.1, 3.5, 1.4, 0.2]]

prediction = pipeline.predict(sample)

print("Prediksi:", prediction)
```

## 13.7 Menyimpan Metadata Model

Selain model, informasi penting juga sebaiknya disimpan.

Contohnya:

- nama model
- versi model
- fitur yang digunakan
- tanggal training
- metric evaluasi

Metadata membantu kita mengetahui bagaimana model tersebut dibuat.

```python
import json

metadata = {
    "model_name": "Iris Logistic Regression",
    "version": "1.0",
    "features": [
        "sepal_length",
        "sepal_width",
        "petal_length",
        "petal_width"
    ],
    "accuracy": 0.95
}

with open("model_metadata.json", "w") as file:
    json.dump(
        metadata,
        file,
        indent=4
    )

print("Metadata berhasil disimpan.")
```

## 13.8 Model sebagai Fungsi Prediction

Dalam aplikasi, kita dapat membuat fungsi sederhana untuk menerima input dan menghasilkan prediction.

```python
import joblib

model = joblib.load(
    "iris_pipeline.pkl"
)

def predict_iris(features):
    prediction = model.predict([features])

    return prediction[0]


result = predict_iris(
    [5.1, 3.5, 1.4, 0.2]
)

print("Hasil:", result)
```

## 13.10 Monitoring Model

Model yang sudah digunakan di production tetap perlu dipantau.

Hal yang dapat dipantau:

- jumlah prediction
- waktu response
- error
- perubahan distribusi data
- akurasi jika label tersedia
- perubahan performa model

Model yang awalnya bagus belum tentu selalu bagus ketika data dunia nyata berubah.

Perubahan pola data disebut data drift.

## 13.11 Model Lifecycle

Secara umum lifecycle model machine learning:

2. Membersihkan data.

4. Evaluasi model.
5. Menyimpan model.
6. Deploy model.
7. Monitoring.
